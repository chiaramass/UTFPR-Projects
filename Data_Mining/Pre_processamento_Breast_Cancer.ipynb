# SimpleImputer

```python
#importar classe simpleimputer
from sklearn.impute import SimpleImputer
```

```python
#faz leitura de base de dados
df = pd.read_csv('/content/breast_cancer_missing.csv')
```

```python
df.isnull().sum()
```

```python
#remover os dados faltantes da classe diagnosis
df.dropna(subset= ['diagnosis'], inplace = True)
df['diagnosis']
```

```python
# retorna nomes das colunas(atributos) com qualquer dados faltante
nan_columns = df.columns[df.isna().any()].tolist()
nan_columns
```

```python
# cria objeto simple imputer
#missing value : indica quale valore deve considerare come mancante
#strategy: indica quale startegia applicare ai dati mancanti.
#In questo caso sostiruirla con la media dei valori presenti (not null) nella stessa colonna
imputer = SimpleImputer(missing_values= np.nan, strategy= 'mean' )
```

```python
#Aplica transformacao
#fit(): L'imputer "impara" i parametri necessari dalla colonna (o dalle colonne) specificata.
#fit() : calcolera la media di ogni colonna senza nan
#transform() : dop aver imparato le medie applica la trasformazione dei NaN per la media degli altri valori della colonna
df[nan_columns] = imputer.fit_transform(df[nan_columns])
```

```python
#verificaare se ci sono ancora dati null
df.isnull().sum()
```

*   StandardScaler : é um pré-processador de dados que transforma os dados de forma que eles tenham uma média de 0 e um desvio padrão de 1. Este é um passo comum na preparação de dados para muitos algoritmos de machine learning, pois pode ajudar a melhorar o desempenho do modelo.

* MinMaxScaler
